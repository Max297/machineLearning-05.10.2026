# Введение в объектно-ориентированное программирование (ООП) на Python

**Класс** — это шаблон (чертёж) для создания объектов. Класс описывает:
- **атрибуты** — данные объекта (например, `balance`, `login`);
- **методы** — функции, которые работают с этими данными (например, `deposit()`, `withdraw()`).

**Объект (экземпляр класса)** — это конкретная «копия», созданная по шаблону класса.

### Ключевые слова:
- `class` — объявление класса;
- `__init__` — специальный метод-конструктор, вызывается при создании объекта;
- `self` — ссылка на сам объект (первый параметр каждого метода класса);
- `super()` — вызов метода родительского класса (используется при наследовании).

## Пример 1. Простой класс `bankAccount`

Разберём класс построчно. Обратите внимание:
- `__init__` — это **конструктор**, он инициализирует атрибуты нового объекта.
- Все атрибуты записываются через `self.<имя>` — так мы «привязываем» их к конкретному объекту.
- Методы объявляются как обычные функции, но **обязательно** с первым параметром `self`.

````python
from datetime import datetime
import random

# Объявляем класс. По соглашению имена классов пишутся в стиле CamelCase.
class bankAccount:
    # __init__ — конструктор. Вызывается автоматически при создании объекта.
    # self — ссылка на создаваемый объект.
    # idNew, balanceNew — параметры, которые пользователь передаёт при создании.
    def __init__(self, idNew, balanceNew):
        # Приведение типов: если баланс передан строкой, превращаем её в число.
        if isinstance(balanceNew, str):
            balanceNew = int(balanceNew)

        # Создаём атрибуты объекта. Теперь у каждого объекта будут свои id и balance.
        self.id = idNew
        self.balance = balanceNew
        self.history = []          # список для хранения истории операций

    # Метод пополнения счёта.
    def deposit(self, money):
        # isinstance — более корректный способ проверки типа, чем type(...) == int
        if isinstance(money, int) or (isinstance(money, str) and money.isdigit()):
            money = int(money)
            if money > 0:
                self.balance += money
                print("Аккаунт пополнен")
                # Вызываем другой метод этого же класса через self
                self.record("Пополнение", money, datetime.now())
            else:
                print("Укажите положительное число")
        else:
            print("Пожалуйста, укажите число")

    # Метод снятия денег
    def withdraw(self, money):
        if isinstance(money, int) or (isinstance(money, str) and money.isdigit()):
            money = int(money)
            if money > 0 and self.balance >= money:
                self.balance -= money
                print("Сняли деньги с аккаунта")
                self.record("Снятие", money, datetime.now())
            else:
                print("Укажите корректную сумму")
        else:
            print("Пожалуйста, укажите число")

    # Вспомогательный метод — записывает операцию в историю
    def record(self, method, money, date):
        operation = {
            "Операция": method,
            "Сумма": money,
            "Дата": date
        }
        self.history.append(operation)

    # Метод-геттер: возвращает текущий баланс
    def getBalance(self):
        return self.balance

    # Метод для вывода истории операций
    def getHistory(self):
        print(self.history)
````

###  Проверим работу класса `bankAccount`

````python
# Создаём объект класса bankAccount. При этом автоматически вызывается __init__.
acc = bankAccount(idNew=123, balanceNew=1000)

# Вызываем методы объекта через точку
acc.deposit(500)        # пополним на 500
acc.withdraw(200)       # снимем 200
print("Баланс:", acc.getBalance())
acc.getHistory()
````

## Пример 2. Наследование — класс `savingsAccount`

**Наследование** позволяет создать новый класс на основе уже существующего.

Синтаксис:
```python
class ChildClass(ParentClass):
    ...
```

- **Родительский класс** (`bankAccount`) — содержит общий функционал.
- **Дочерний класс** (`savingsAccount`) — наследует всё от родителя и может добавлять своё.

`super().__init__(...)` — вызываем конструктор родителя, чтобы не дублировать код инициализации.

````python
# savingsAccount наследуется от bankAccount
class savingsAccount(bankAccount):
    # Конструктор дочернего класса
    def __init__(self, id, balance, interestRate=0.1):
        # super() возвращает объект родительского класса.
        # Через него мы вызываем __init__ родителя, передавая id и balance.
        super().__init__(id, balance)

        # Добавляем новый атрибут, которого не было у родителя
        self.interestRate = interestRate

    # Новый метод, специфичный только для сберегательного счёта
    def calculateInterest(self):
        newBalance = self.balance + self.balance * self.interestRate
        print(f"На счет пришли накопления по вкладу, новый баланс: {newBalance}")
        self.balance = newBalance
````

````python
# Создаём сберегательный счёт с процентной ставкой 5%
sav = savingsAccount(id=456, balance=2000, interestRate=0.05)
sav.deposit(500)             # метод унаследован от bankAccount
sav.calculateInterest()      # собственный метод savingsAccount
print("Баланс:", sav.getBalance())
````

# Пример: форум

Постановка задачи:
1. **User** (Пользователь) — логин, пароль, ID.
2. **Comment** (Комментарий) — текст, дата, автор, ID, ответ на другой комментарий.
3. **Post** (Пост) — заголовок, текст, автор, дата, ID, набор комментариев.
4. **Forum** (Форум) — хранит пользователей и посты.

Здесь мы увидим, как классы **взаимодействуют друг с другом**: `Forum` создаёт объекты `User` и `Post`, а `Post` создаёт объекты `Comment`.

````python
class User:
    def __init__(self, newLogin, newPassword):
        self.login = newLogin
        self.password = newPassword
        # ID генерируем случайно — это уникальный идентификатор пользователя
        self.id = random.randint(0, 10**20)
````

````python
class Comment:
    def __init__(self, text, author, responseId):
        self.text = text
        self.author = author
        self.responseId = responseId   # ID комментария, на который отвечают
        self.date = datetime.now()
        self.id = random.randint(0, 10**20)
````

````python
class Post:
    def __init__(self, title, text, author):
        self.title = title
        self.text = text
        self.author = author
        self.date = datetime.now()
        self.id = random.randint(0, 10**20)
        # Комментарии храним в словаре: ключ — ID комментария, значение — объект Comment
        self.comments = {}

    def addComment(self, text, author, responseId):
        newComment = Comment(text, author, responseId)
        self.comments[newComment.id] = newComment

    def deleteComment(self, commentId):
        # pop удаляет элемент по ключу; если ключа нет — ничего не делает
        self.comments.pop(commentId, None)
````

````python
class Forum:
    def __init__(self):
        self.users = {}     # {id: User}
        self.posts = {}     # {id: Post}

    def addUser(self, login, password):
        newUser = User(login, password)
        self.users[newUser.id] = newUser
        return newUser

    def addPost(self, title, text, author):
        newPost = Post(title, text, author)
        self.posts[newPost.id] = newPost
        return newPost

    def createComment(self, text, author, postId, responseId=-1):
        # Находим пост по ID и добавляем в него комментарий
        self.posts[postId].addComment(text, author, responseId)

    def deleteComment(self, postId, commentId, userId):
        # Проверяем права: удалить комментарий может только его автор
        if userId == self.posts[postId].comments[commentId].author:
            self.posts[postId].deleteComment(commentId)
        else:
            print(f"У пользователя {userId} нет прав для удаления комментария {commentId}")

    def deletePost(self, postId, userId):
        if userId == self.posts[postId].author:
            self.posts.pop(postId)
        else:
            print(f"У пользователя {userId} нет прав для удаления поста {postId}")
````

 Тестируем

````python
forum = Forum()

# Регистрация пользователей
alice = forum.addUser("alice", "12345")
bob = forum.addUser("bob", "qwerty")

# Alice создаёт пост
post = forum.addPost("Привет, форум!", "Мой первый пост.", alice.id)

# Bob оставляет комментарий
forum.createComment("Отличный пост!", bob.id, post.id)

print("Постов на форуме:", len(forum.posts))
print("Комментариев к посту:", len(forum.posts[post.id].comments))
````

## Краткая шпаргалка

| Понятие | Что это | Пример |
|---|---|---|
| Класс | Шаблон объекта | `class User: ...` |
| Объект | Экземпляр класса | `u = User("alice", "123")` |
| `__init__` | Конструктор | вызывается при `User(...)` |
| `self` | Ссылка на объект | используется во всех методах |
| Атрибут | Данные объекта | `self.balance` |
| Метод | Функция класса | `def deposit(self, money): ...` |
| Наследование | Расширение класса | `class Child(Parent): ...` |
| `super()` | Вызов родителя | `super().__init__(...)` |
