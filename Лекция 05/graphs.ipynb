```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

Пример созданий линейного графика

Во всех видах графиков нам нужно сначала создать "рисунок" на котором будет строиться график plt.figure(figsize=(10,6))

В линейном графике нужно указать x и y.

В большенстве случаяев мы можем строить на одном рисунке несколько графиков одного типа

```python
months=["Янв","Фев","Мар","Апр","Май","Июн"]
sales_1=[150,180,200,170,220,250]
sales_2=[120,160,280,210,290,350]

plt.figure(figsize=(10,6))
plt.plot(months, sales_1,marker="o",linewidth=3,color='black')
plt.plot(months, sales_2,marker="o",linewidth=3,color='red')

plt.title("Продажи по месяцам")
plt.xlabel("Месяц")
plt.ylabel("Продажи")

plt.show()
```

Столбчаттая диограмма

```python
cities=["Москва","Спб","Новосибирск","Екатеринбург","Казань"]
popSize=[12.6,5.4,1.6,1.5,1.3]

plt.figure(figsize=(10,6))
bars=plt.bar(cities,popSize,edgecolor='black')

for bar in bars:
  height = bar.get_height()

  text=str(height)+" млн"
  plt.text(bar.get_x() + bar.get_width()/4,height,text)

plt.title("Население городов")
plt.ylabel("Население в млн")
plt.xlabel("Город")

plt.show()
```

Точечный график

```python
np.random.uniform(1,5,50)
```

```python
hours=np.random.uniform(1,10,50)
scores=30+5*hours + np.random.normal(0,10,50)
age = np.random.uniform(1,5,50)

plt.figure(figsize=(10,6))
plt.scatter(hours, scores ,
            c=age,
            s=60,
            alpha=0.5,
            edgecolor="black")

plt.title("Зависимость оценки от вермени обучения")
plt.xlabel("Часы обучения")
plt.ylabel("Оценка")
plt.show()
```

Гистограмма

```python
salaries=np.random.uniform(60000,150000,1000)
plt.figure(figsize=(10,6))
plt.hist(
    salaries,
    bins=20,
    color="blue",
    edgecolor="black",
    alpha=0.7
)

plt.title("Распределение зарплат")
plt.xlabel("Зарплата")
plt.ylabel("Частота")
plt.show()
```

Круговая диограмма

```python
cats=["Продукт 1","Продукт 2","Продукт 3","Продукт 4"]
sizes=[35,25,22,18]
colors=["red","blue","green","yellow"]

plt.figure(figsize=(10,6))
plt.pie(sizes,labels=cats,colors=colors,shadow=True,startangle=90)

plt.show()
```

Пример

```python
data={
    "День":["Пн","Вт","Ср","Чт","Пт"],
    "Товар 1":[30,35,40,38,50],
    "Товар 2":[45,50,55,60,75],
    "Товар 3":[40,42,45,50,65]
}

df=pd.DataFrame(data)
```

```python
df
```

```python
weekly_total=df[["Товар 1","Товар 2","Товар 3"]].sum()
```

```python
weekly_total
```

```python
fig,axes = plt.subplots(1,3,figsize=(18,5))

ax1=axes[0]

ax1.plot(df["День"],df["Товар 1"],marker="o",color="red",label="Товар 1")
ax1.plot(df["День"],df["Товар 2"],marker="s",color="blue",label="Товар 2")
ax1.plot(df["День"],df["Товар 3"],marker="^",color="green",label="Товар 3")

ax1.set_title("Динамика продаж товаров на неделе")


ax2=axes[1]

bars=ax2.bar(weekly_total.index,weekly_total.values,color=["red","blue","green"])

for bar in bars:
  height = bar.get_height()

  text=str(height)+" продано"
  ax2.text(bar.get_x() + bar.get_width()/4,height,text)



ax3=axes[2]

ax3.pie(weekly_total,labels=weekly_total.index,colors=["red","blue","green"])



plt.show()
```
