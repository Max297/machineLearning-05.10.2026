Срезы - способ получить часть массива

item[START:STOP:STEP] - берёт срез от номера START, до STOP (не включая его), с шагом STEP. По умолчанию START = 0, STOP = длине объекта, STEP = 1. Соответственно, какие-нибудь (а возможно, и все) параметры могут быть опущены.

Срезы могут быть отрицательными

Примеры объявления функций

```python

def add(x, y):
  """Сложение двух чисел."""
  return x + y

def subtract(x, y):
  """Вычитание двух чисел."""
  return x - y

def multiply(x, y):
  """Умножение двух чисел."""
  return x * y

def divide(x, y):
  """Деление двух чисел."""
  if y == 0:
    return "Деление на ноль!"
  return x / y

def power(x, y):
  """Возведение числа в степень."""
  return x ** y

def square_root(x):
  """Вычисление квадратного корня."""
  if x < 0:
    return "Квадратный корень из отрицательного числа не существует!"
  return x ** 0.5
```

Игра "Угадай число"

```python
def guessTheNumber(x):
  numberGuessed=False
  while not numberGuessed:
    userNumber = int(input("Угадайте чисило: "))
    if userNumber > x :
      print("Не угадал! Загаданное число меньше")
    elif userNumber < x:
      print("Не угадал! Загаданное число больше")
    else:
      print("Вы победили")
      numberGuessed = True
```

Шаг 1, сортировка 2 66->234

Шаг 1, сортировка 3 234->723

```python
unsortedArray=[10,-2,6,10,66,1,-5,0,234,723,4512,-632,11,97,0]
```

```python
def bubbleSort(array,sortType="возрастение"):
  arrayLength = len(array)

  for i in range(0,arrayLength):
    for j in range (i+1,arrayLength):
      if(sortType == "возрастение"):
        if(array[i] > array[j]):
            vrem = array[i]
            array[i]=array[j]
            array[j]=vrem
      elif(sortType == "убывание"):
        if(array[i] < array[j]):
          vrem = array[i]
          array[i]=array[j]
          array[j]=vrem
      else:
        print("Неизвестный тип сортировки")
        return array
  return array
```
